```python
import numpy as np
import pandas as pd
from elpv_dataset.utils import load_dataset
images, proba, types = load_dataset()
```

```python
print(images.dtype)
print(proba.dtype)
print(types.dtype)
```

```python
print(f"Total Images Loaded: {images.shape[0]}")
print(f"Image Resolution: {images.shape[1]}x{images.shape[2]} pixels")
```

```python
df = pd.DataFrame({
    'probability': proba,
    'cell_type': types
})

df['condition'] = np.where(df['probability'] == 0.0, 'Functional', 'Defective')

print("--- Cell Condition Distribution ---")
condition_counts = df['condition'].value_counts()
condition_pct = df['condition'].value_counts(normalize=True) * 100
for idx in condition_counts.index:
    print(f"{idx}: {condition_counts[idx]} cells ({condition_pct[idx]:.2f}%)")

print("\n--- Crystalline Type Distribution ---")
type_counts = df['cell_type'].value_counts()
type_pct = df['cell_type'].value_counts(normalize=True) * 100
for idx in type_counts.index:
    print(f"{idx.capitalize()}crystalline: {type_counts[idx]} cells ({type_pct[idx]:.2f}%)")

print("\n--- Combined Distribution Matrix ---")
matrix = pd.crosstab(df['cell_type'], df['condition'])
print(matrix)
```
